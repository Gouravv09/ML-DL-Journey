```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "I love pizza",
    "Pizza is the best",
    "I love pasta",
    "Pasta  is great"
]
```

```python
bow = CountVectorizer()
```

```python
X = bow.fit_transform(documents)
```

```python
print("Vocabulary: ", bow.get_feature_names_out())
print("\nBow Matrix:\n", X.toarray())
```

```python
ngram = CountVectorizer(ngram_range = (2, 2))
```

```python
X = ngram.fit_transform(documents)
```

```python
print("Vocabulary: ", ngram.get_feature_names_out())
print("\nBow Matrix:\n", X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer()
```

```python
X = tfidf.fit_transform(documents)
```

```python
print("Vocabulary: ", tfidf.get_feature_names_out())
print("\nBow Matrix:\n", X.toarray())
```
